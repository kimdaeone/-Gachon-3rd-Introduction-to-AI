# **CNN, RNN, LSTM의 이해**

# **서론**

**대표적인 딥러닝 세가지 아키텍처 **

CNN, RNN, LSTM은 다루고자 하는 데이터의 차원적 속성(공간적 인접성 및 시간적 연속성)에 따라 서로 다른 최적화 전략을 채택하고 있어, 이에 대해 메커니즘과 비교 분석을 해보고자 한다




---

# **CNN (Convolutional Neural Network) : 공간 정보**

이미지 데이터는 각 픽셀의 물리적 위치가 매우 중요한 공간적인 특성(Spatial Features)을 지닌다

>코의 왼쪽에 눈이 위치한다는 정보처럼 픽셀 간의 인접 관계는 형상을 인지하는 단서

그러나 이미지를 한 줄의 1차원 벡터로 일렬로 펼쳐서 입력하는 경우 , 인접한 픽셀 간의 물리적 위치 정보와 상관관계가 달라지며, 모델은 이것을 이해하지 못하여 왜곡하는 현상이 발생한다


### **합성곱(Convolution)과 정보 처리**

CNN은 이미지를 평탄화하지 않고, 원래의 2차원 또는 3차원 격자 구조를 유지한 채 합성곱(Convolution) 연산을 적용하여 이러한 한계를 극복한다







**필터(Kernel)와 가중치 공유(Weight Sharing):** :

필터는 이미지 전체를 일정한 보폭(Stride)으로 미끄러지듯 이동하면서 겹치는 영역의 입력값과 가중치를 원소별로 곱하고 더하여 새로운 특징 지도(Feature Map)를 생성한다

완전연결층에서는 모든 입력 노드가 각각의 가중치를 독립적으로 가졌던 것과 달리, CNN은 작은 필터를 전체 이미지에 공유하는 가중치 공유를 채택한다

이것은 모델의 가중치 수치를 줄이는 방법이 되어 과적합(Overfiting)을 방지하고 학습 효율성을 높인다

**패딩(Padding):** :

필터 연산을 거듭할수록 이미지의 외각 데이터 연산 참여 횟수가 내부보다 상대적으로 적어져 출력 크기가 줄어든다

주로 가장자리에  특정 값(0을 보통 기입)을 채워 넣는 제로 패딩(Zero Padding)을 수행하여 데이터 손실을 줄이고 출력 크기를 입력과 동일하게 유지한다


**풀링(Pooling):** :
- 최대 풀링(Max Pooling):
영역 내에서 가장 큰 값만 추출하여 노이즈를 억제하고 대표 특징을 요약한다

- 평균 풀링(Average Pooling):
영역 내 값들의 평균을 계산하여 전체적으로 순한 특징들을 보존한다

- 이동 불변성(Translation Invariance): 풀링을 거친 특징 지도는 대상의 위치가 조금씩 달라지더라도 핵심 특징이 동일한 구역에 압축되므로, 이미지 내에서 물체가 어디에 있든 안정적으로 인식할 수 있게 돕는다


# **RNN (Recurrent Neural Network) : 시간적 정보**

CNN이 멈춰 있는 공간의 형태를 분석한다면, RNN은 문장, 음성 신호, 주가 그래프 등 선후 관계와 흐름이 본질이라고 할 수 있는 시계열(Sequens) 데이터를 다룬다

>데이터의 길이가 고정되어 있지 않고, 선행 데이터가 후행 데이터에 영향을 미치는 구조다

이전 시점(t-1)의 은닉 상태(Hidden State)에서 생성된 출력값을 다시 현재 시점(t)의 입력과 결합하여 순환시키는 피드백 루프(Loop)를 가진다

> 은닉 상태는 매 시점마다 출력 계산에 영향을 주는 장치의 역할을 한다

### **치명적인 한계**

문장의 길이가 길어지는 등의 이유로 계산 시간이 증가하면, 역전파 알고리즘에 의해
 체인 룰 기반의 연산이 지속한 시간만큼 반복 적용이 된다.

 >체인 룰은 손실을 최소화 하기 위해 가중치를 계속 업데이트 하는 규칙이다

과정에서 미분의 값이 1보다 작을 경우, 곱셈 연산이 누적되어 기울기가 기하급수적으로 0에 가깝게 수렴하는 기울기 소실(vanishing gradient) 현상이 발생한다

소실로 인해 초기 정보들의 영향력이 옅어지면서 현재 시점의 예측에 전혀 반영되지 못하는 문제에 직면하게 된다






# **LSTM (Long Short-Term Memory): 장기 기억의 보존**

**장기 기억 셀 상태 :**

LSTM은 RNN의 가장 치명적인 약점인 기울기 소실을 극복하기 위해 제안된 신경망이다

단기 정보 처리에 치중된 은닉 상태(h_t) 외에, 정보의 훼손 없이 네트워크의 끝까지 안전하게 핵심 정보가 흐를 수 있는 고속도로 역할의 셀 상태(Cell State, (C_t)) 통로를 따로 구축한다

>셀 상태는 주로 곱셈 연산이 아닌 선형 덧셈 연산을 통해 가중치가 업데이트되므로, 오랜 시간이 지나도 기울기가 유실되지 않고 원점까지 원활하게 오차가 전파될 수 있다



- 망각 게이트 :
이전 은닉 상태(h_{t-1})와 현재 입력(x_t)을 판단하여 과거 셀 상태(C_{t-1}) 중 소멸시켜야 할 필요 없는 기억의 양을 조절한다

출력 값이 0에 가까울수록 과거의 데이터를 잊고, 1에 가까울수록 데이터를 유지한다

- 입력 게이트 :
현재 입력된 정보(x) 중 가장 가치 있는 것을 선택한다

하이퍼볼릭 탄젠트(tanh) 함수가 활성화된 현재의 새로운 정보 후보 라인업(cadinate, (Ct)을 만든 뒤, 두 값을 곱하여 최종적으로 장기 기억인 셀 상태에 더하여 업데이트한다

- 출력 게이트 :
업데이트 된 셀 상태(Ct)의 장기 기억 중 현재의 예측과 은닉 상태(ht)에 직접적으로 관여할 중요한 정보의 비율을 조절한다

>LSTM의 핵심 >>
직전의 은닉 상태와 현재 입력을 활용하여, 이전 장기 기억 벨트(셀 상태)에서 잊어버릴 것은 망각 게이트로 지워버리고, 새로 채워 넣을 핵심은 입력 게이트로 더하여, 최적의 장기 기억 흐름을 실시간으로 유지하는 형태이다

# **결론**

**CNN :**

데이터의 형태적 왜곡 없이 특징을 공유, 추출하는 CNN은 시각 정보 분석 분야에서 강력하다

**RNN :**

선 후 관계의 꼬리를 물고 데이터를 해석하는 RNN은 언어와 시간적 차이의 개념을 활용했다

**LSTM :**

RNN의 기억  상실이라는 한계를 3가지 게이트 형태로 해결한 LSTM은 현재를 예측하고 미래를 학습하는 현대 인공지능의 초식이라고 할 수 있다
